Esse código resolve a questão de salvar o arquivo que está no catálogo para a pasta criada no workspace

```python
import os

CATALOGO = "data-growth-analytic"
ESQUEMA = "gold"

# 1. Busca dinamicamente a lista de tabelas que você tem
tabelas_df = spark.sql(f"SHOW TABLES IN `{CATALOGO}`.{ESQUEMA}")
lista_tabelas = [row["tableName"] for row in tabelas_df.collect()]

# 2. Itera sobre cada tabela encontrada
for nome_tabela in lista_tabelas:
    print(f"📦 Convertendo e salvando: {nome_tabela}...")
    
    # Lê do Unity Catalog usando Spark
    caminho_origem_uc = f"`{CATALOGO}`.{ESQUEMA}.`{nome_tabela}`"
    df_spark = spark.table(caminho_origem_uc)
    
    # Define o nome do arquivo CSV final baseado no nome da tabela
    nome_arquivo_csv = f"{nome_tabela}.csv"
    
    # 3. Transforma em Pandas e grava localmente na pasta atual do seu Workspace
    df_spark.toPandas().to_csv(
        nome_arquivo_csv,
        index=False,
        encoding="utf-8-sig"
    )
    print(f"   -> [OK] Criado arquivo: {nome_arquivo_csv}")

print("\n💥 Todos os arquivos CSV foram gerados com sucesso na pasta do seu Workspace!")
```
